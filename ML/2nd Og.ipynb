```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df2["cgpa"],df2["iq"],c=df2["placement"])
plt.xlabel("CGPA")
plt.ylabel("IQ")
plt.title("CGPA v/s IQ")
plt.show()
```

```python
plt.hist(df2["cgpa"], bins=10, color='#ff00ff', edgecolor='black')
plt.xlabel("CGPA")
plt.ylabel("Frequency")
plt.title("Distribution of CGPA")
plt.show()
```

```python
import seaborn as sns

# Box plot for CGPA grouped by placement
df2.boxplot(column='cgpa', by='placement')
plt.title('CGPA by Placement')
plt.suptitle('')
plt.xlabel('Placement')
plt.ylabel('CGPA')
plt.show()

# Pairplot for all features
sns.pairplot(df2, hue='placement')
plt.show()

# Violin plot for IQ grouped by placement
sns.violinplot(x='placement', y='iq', data=df2)
plt.title('IQ Distribution by Placement')
plt.xlabel('Placement')
plt.ylabel('IQ')
plt.show()
```
