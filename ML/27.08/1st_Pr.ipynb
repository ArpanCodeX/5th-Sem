```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split as tts
from sklearn.neighbors import KNeighborsRegressor as kn
from sklearn.metrics import mean_squared_error as msq
url="https://raw.githubusercontent.com/mubaris/potential-enigma/master/headbrain.csv"
data=pd.read_csv(url)
print("First few rows of the dataset :-")
print(data.head())
X=data['Head Size(cm^3)'].values.reshape(-1,1)
Y=data['Brain Weight(grams)'].values.reshape(-1,1)
X_train,X_test,Y_train,Y_test=tts(X,Y,test_size=0.3,random_state=42)
knn=kn(n_neighbors=3)
knn.fit(X_train,Y_train)
Y_pred=knn.predict(X_test)
mse=msq(Y_test,Y_pred)
print(f"Mean Squared Error:{mse}")
comparison=pd.DataFrame({'Actual':Y_test.ravel(),'Predicted':Y_pred.ravel()})
print("\n comparison of Actual and Predicted values :-")
print(comparison.head())
plt.scatter(X_test,Y_test,color='lightgreen',label='Actual')
plt.scatter(X_test,Y_pred,color='lightcoral',label='Predicted')
plt.title('KNN Regression-Actual vs Predicted')
plt.xlabel('Head size(cm^3)')
plt.ylabel('Brain weight(grams)')
plt.legend()
plt.show()
```
