## Prime number in range
numpy,pandas, seaborn , matplotlib, scikitlearn, scipy, mlxtend


```python
l = int(input("Enter lower range  : "))
h = int(input("Enter higher range : "))

prime = []
for i in range(l, h + 1):
    if i > 1:
        for j in range(2, int(i ** 0.5) + 1):
            if i % j == 0:
                break
        else:
            prime.append(i)

print(f"Prime numbers in range ({l},{h}) : {prime}")
```
