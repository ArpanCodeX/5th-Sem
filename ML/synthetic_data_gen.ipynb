# **Write a program to plot the first feature and second feature of dataset Iris.**

```python
import mglearn
import matplotlib.pyplot as plt
```

```python
X, y = mglearn.datasets.make_forge()

mglearn.discrete_scatter(X[:, 0], X[:, 1], y)
plt.legend(["class 0", "class 1"], loc=4)
plt.xlabel("first feature:")
plt.ylabel("2nd feature:")
print(X, y)
print("X.shape:{}".format(X.shape))
plt.show()
```

# **Write a program to plot features and target of Iris dataset.**

```python
import mglearn
import matplotlib.pyplot as plt
```

```python
x, y = mglearn.datasets.make_wave(n_samples=40)

plt.plot(x, y, 'o')
plt.ylim(-3, 3)
plt.xlabel("Features")
plt.ylabel("Target")
plt.show()
```
